```python
# Prompt the user for input and store it as a variable num_days
num_days = int(input("Please enter the number of days: "))
# calculate the number of seconds in a user's input number of days by multiplying 24 hours by 60 minutes by 60 seconds
# and store it in a variable total_seconds
total_seconds = num_days * 24 * 60 * 60

# print out the
print(f"The total number of seconds in {num_days} is {total_seconds} seconds")
```

```python
import math
r = float(input("Enter the radius of the sphere: ")) # prompts user for radius (float)
v = (4/3) * math.pi * r ** 3 # calculates the volume using the formula
print(f"The sphere's volume is {v}") # prints out the answer

```

```python
def area(side): # function to calculate area  by raising the param to power 2
  return side ** 2

def perimeter(side): # function to calculate perimeter by multiplying param length by 4
  return side * 4

length = float(input("Enter the side of the square: "))
area_val = area(length) # storing the return of calling area function
perimeter_val = perimeter(length) # storing the return of calling the perimeter function
print(f"The area is {area_val} while the perimeter is {perimeter_val}") # printing both values
```

```python
def isUpper(chr): # function to check store the result of checking if character is in upper case by being between A and Z
  if 'A' <= chr <= 'Z':
    return True # True if between A and Z
  return False # False otherwise

inp = input("Enter the character: ")
if isUpper(inp):
  print("The character is upper case")
else:
  print("The character is in lower case")
```

```python
# initialize the variables
x = 0
y = 20

# loop as long as y is greater than 6 (repeat until this condition fails)
while y>6:
  # y = y - 4
  y-=4
  # x = x + 2/y
  x += 2/y
print(f"The value of x is {x}")
```

```python
import array

ls = [] # initialize an empty list
ls = array.array('i', []) # convert to an array of integers
for i in range(5): # loop 5 times
  inp = int(input(f"Enter value {i+1}: ")) # prompt user for values each iteration
  ls.append(inp) # append the value in the array
sum = 0 # declare a sum variable
for n in ls: # loop through the numbers in the array
  sum+=n # add each of them
print(f"The average is {sum/5}") # print the average

```
